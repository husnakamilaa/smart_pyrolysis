## encoding PS (partikel size)

```python
for col in ['Solid phase', 'Liquid phase', 'Gas phase']:
    df_clean[col] = df_clean[col].astype(str).str.extract(r'(\d+\.?\d*)')[0]
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

def clean_ps_to_float(val):
    if pd.isnull(val) or val == 'nan':
        return np.nan

    val_str = str(val).strip()

    # Jika mendeteksi pola rentang dengan tanda strip (misal '0.5-0.6')
    if '-' in val_str:
        try:
            parts = val_str.split('-')
            # Hitung nilai tengahnya: (0.5 + 0.6) / 2 = 0.55
            return (float(parts[0]) + float(parts[1])) / 2.0
        except:
            return np.nan

    # Jika angka biasa, langsung ubah ke float
    try:
        return float(val_str)
    except:
        return np.nan

# Terapkan fungsi pembersihan pada kolom PS
df_clean['PS_numeric'] = df_clean['PS'].apply(clean_ps_to_float)

print("Tipe data PS sekarang:", df_clean['PS_numeric'].dtype)
print("Jumlah missing value di PS sekarang:", df_clean['PS_numeric'].isnull().sum())
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

# 1. Inisialisasi DataFrame
df_prox = df_clean.copy()
prox_cols = ['M', 'Ash ', 'VM', 'FC']
all_features = ['M', 'Ash ', 'VM', 'FC', 'C', 'H', 'O', 'N', 'HR', 'FR', 'PS']

# 2. Masking untuk memisahkan 18 baris yang kosong total proksimatnya
empat_kosong_mask = df_prox[prox_cols].isnull().all(axis=1)

df_normal = df_prox[~empat_kosong_mask].copy()  # Kelompok Utama
df_blanks = df_prox[empat_kosong_mask].copy()  # Kelompok 18 Baris

print(f"Memproses {df_normal.shape[0]} baris Kelompok Utama...")

# --- LANGKAH A: Eksekusi Kelompok Utama dengan Rumus Sisa 100% ---
median_prox = {col: df_normal[col].median() for col in prox_cols}

for idx, row in df_normal.iterrows():
    null_cols = [col for col in prox_cols if pd.isnull(row[col])]
    if len(null_cols) == 0:
        continue

    dependent_col = null_cols[-1]
    cols_to_fill_with_median = null_cols[:-1]

    for col in cols_to_fill_with_median:
        df_normal.at[idx, col] = median_prox[col]

    current_sum = sum(df_normal.at[idx, c] for c in prox_cols if c != dependent_col)
    calculated_val = 100.0 - current_sum
    df_normal.at[idx, dependent_col] = max(0.0, calculated_val)


# --- LANGKAH B: Eksekusi Kelompok 18 Baris Menggunakan KNN Imputer ---
print(f"Memproses {df_blanks.shape[0]} baris menggunakan KNN Imputer...")

# Gabungkan sementara untuk memberikan konteks fitur Ultimat (C,H,O,N, dll) ke KNN
# Ambil seluruh data numerik yang sudah terisi untuk membantu KNN belajar kemiripan
df_for_knn = df_prox[all_features].copy()

# Panggil KNN Imputer (melihat 5 tetangga terdekat berdasarkan fitur lain yang tersedia)
knn_imputer = KNNImputer(n_neighbors=5)
df_knn_imputed = pd.DataFrame(knn_imputer.fit_transform(df_for_knn), columns=all_features, index=df_prox.index)

# Ambil hasil imputasi proksimat khusus untuk 18 baris tadi
for col in prox_cols:
    df_blanks[col] = df_knn_imputed.loc[empat_kosong_mask, col]

# --- LANGKAH C: Normalisasi Proporsional Hasil KNN agar Tepat 100% ---
# Karena KNN menghasilkan angka estimasi, totalnya mungkin 98% atau 102%. Kita paksa ke 100%.
for idx, row in df_blanks.iterrows():
    total_knn = sum(df_blanks.at[idx, col] for col in prox_cols)
    if total_knn > 0:
        for col in prox_cols:
            # Rumus standarisasi proporsional: (Nilai / Total_Skarang) * 100
            df_blanks.at[idx, col] = (df_blanks.at[idx, col] / total_knn) * 100.0


# --- LANGKAH D: Gabungkan Kembali Data ---
df_final_prox = pd.concat([df_normal, df_blanks]).sort_index()
print("\nProses Imputasi Gabungan Selesai!")
```

```python
df_final_prox['Total_Proximate'] = df_final_prox[prox_cols].sum(axis=1)

print("=== Hasil Validasi Akhir (Constraint + KNN) ===")
print("Rata-rata total proksimat seluruh data :", round(df_final_prox['Total_Proximate'].mean(), 2))
print("Nilai total paling minimum di data     :", round(df_final_prox['Total_Proximate'].min(), 2))
print("Nilai total paling maksimum di data     :", round(df_final_prox['Total_Proximate'].max(), 2))
print("\nMissing value tersisa pada proksimat   :\n", df_final_prox[prox_cols].isnull().sum())

# Hapus kolom helper total
df_final_prox.drop(columns=['Total_Proximate'], inplace=True)
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

df_ult = df_final_prox.copy()
ult_cols = ['C', 'H', 'O', 'N']
all_features_ult = ['M', 'Ash ', 'VM', 'FC', 'C', 'H', 'O', 'N', 'HR', 'FR', 'PS']

# --- LANGKAH 1: KOSONGKAN DATA YANG MERUSAK INFORMASI (ANOMALI) ---
# Hitung total sementara untuk deteksi
df_ult['Total_Sement'] = df_ult[ult_cols].sum(axis=1)

# Batas aman total ultimat laboratorium (85% - 105%)
# Jika di luar ini, kita paksa nilainya menjadi NaN agar ditebak ulang oleh KNN
anomali_mask = (df_ult['Total_Sement'] < 85.0) | (df_ult['Total_Sement'] > 105.0)
print(f"Mengosongkan {anomali_mask.sum()} baris data ultimat yang terdeteksi anomali/rusak...")

df_ult.loc[anomali_mask, ult_cols] = np.nan
df_ult.drop(columns=['Total_Sement'], inplace=True)


# --- LANGKAH 2: FILTERING SEPERTI BIASA ---
# Masking untuk mencari baris yang kosong murni (termasuk data anomali yang baru kita kosongkan)
ult_kosong_semua = df_ult[ult_cols].isnull().all(axis=1)

df_ult_normal = df_ult[~ult_kosong_semua].copy()  # Data yang valid / bolong acak sedikit
df_ult_blanks = df_ult[ult_kosong_semua].copy()   # Data yang kosong total + bekas anomali

print(f"Memproses {df_ult_normal.shape[0]} baris Kelompok Normal...")


# --- LANGKAH 3: ISI DATA BOLONG ACAK PADA KELOMPOK NORMAL (RUMUS SISA 100%) ---
median_ult = {col: df_ult_normal[col].median() for col in ult_cols}

for idx, row in df_ult_normal.iterrows():
    null_cols_ult = [col for col in ult_cols if pd.isnull(row[col])]
    if len(null_cols_ult) == 0:
        continue

    dependent_col_ult = null_cols_ult[-1]
    cols_to_fill_with_median_ult = null_cols_ult[:-1]

    for col in cols_to_fill_with_median_ult:
        df_ult_normal.at[idx, col] = median_ult[col]

    current_sum_ult = sum(df_ult_normal.at[idx, c] for c in ult_cols if c != dependent_col_ult)
    calculated_val_ult = 100.0 - current_sum_ult
    df_ult_normal.at[idx, dependent_col_ult] = max(0.0, calculated_val_ult)


# --- LANGKAH 4: ISI KELOMPOK KOSONG TOTAL & BEKAS ANOMALI PAKAI KNN ---
print(f"Memproses {df_ult_blanks.shape[0]} baris menggunakan KNN Imputer...")

df_for_knn_ult = df_ult[all_features_ult].copy()
knn_imputer_ult = KNNImputer(n_neighbors=5)
df_knn_imputed_ult = pd.DataFrame(knn_imputer_ult.fit_transform(df_for_knn_ult), columns=all_features_ult, index=df_ult.index)

for col in ult_cols:
    df_ult_blanks[col] = df_knn_imputed_ult.loc[ult_kosong_semua, col]


# --- LANGKAH 5: GABUNGKAN KEMBALI & NORMALISASI PROPORSIONAL BIAR ADIL ---
df_final_ult = pd.concat([df_ult_normal, df_ult_blanks]).sort_index()

print("Menjalankan penyetaraan proporsional akhir menuju 100% murni...")
for col in ult_cols:
    df_final_ult[col] = df_final_ult[col].clip(lower=0.0)

total_ult_per_baris = df_final_ult[ult_cols].sum(axis=1)

for idx, row in df_final_ult.iterrows():
    t_baris = total_ult_per_baris.loc[idx]
    if t_baris > 0:
        for col in ult_cols:
            df_final_ult.at[idx, col] = (df_final_ult.at[idx, col] / t_baris) * 100.0


# --- LANGKAH 6: SINKRONISASI DESIMAL KOMPUTER ---
for idx, row in df_final_ult.iterrows():
    total_akhir_ult = sum(df_final_ult.at[idx, col] for col in ult_cols)
    if total_akhir_ult != 100.0:
        komponen_terbesar_ult = df_final_ult.loc[idx, ult_cols].astype(float).idxmax()
        selisih_ult = 100.0 - total_akhir_ult
        df_final_ult.at[idx, komponen_terbesar_ult] += selisih_ult

print("\n=== DATA ULTIMAT BERSIH DARI ANOMALI DAN TERKUNCI 100% ===")
```

```python
import pandas as pd
import numpy as np
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler

# 1. Inisialisasi data dari tahap sebelumnya
df_investigasi = df_final_ult.copy()
kolom_spesies = 'Biomass species'

# 2. Tampilkan data yang memiliki PS = 150 dan 175
print("=== DATA ANOMALI PS (150 & 175) ===")
data_anomali = df_investigasi[df_investigasi['PS'].isin([150.0, 175.0])]
display(data_anomali[[kolom_spesies, 'M', 'Ash ', 'VM', 'FC', 'FT', 'PS']])

# --- PROSES PENCARIAN DATA KEMBAR / MIRIP ---

# Fitur yang digunakan untuk mengukur kemiripan (Sidik Jari Biomassa)
fitur_kemiripan = ['M', 'Ash ', 'VM', 'FC', 'FT']

# Kita standarisasi skala fitur kemiripan dulu agar perhitungan jarak adil
scaler_cek = StandardScaler()
X_scaled = scaler_cek.fit_transform(df_investigasi[fitur_kemiripan])

# Buat dataframe scaled dengan index yang sama agar mudah dicocokkan
df_scaled = pd.DataFrame(X_scaled, index=df_investigasi.index)

# Latih model Nearest Neighbors untuk mencari 5 tetangga terdekat
# n_neighbors=6 karena tetangga pertama (indeks ke-0) pasti dirinya sendiri
nn = NearestNeighbors(n_neighbors=6, metric='euclidean')
nn.fit(df_scaled)

# Loop untuk mencari dan menampilkan tetangga mirip untuk setiap baris anomali
for idx in data_anomali.index:
    ps_val = data_anomali.loc[idx, 'PS']
    print(f"\n==================================================================")
    print(f"5 DATA YANG PALING MIRIP DENGAN INDEKS {idx} (PS = {ps_val} mm)")
    print(f"==================================================================")

    # Ambil koordinat scaled dari data anomali ini
    data_point_scaled = df_scaled.loc[[idx]]

    # Cari jarak dan indeks tetangganya
    distances, indices = nn.kneighbors(data_point_scaled)

    # Ambil indeks tetangga (abaikan indeks pertama karena itu dirinya sendiri)
    tetangga_indices = indices[0][1:]

    # Tampilkan 5 data pembanding terdekat
    df_tetangga = df_investigasi.loc[tetangga_indices, [kolom_spesies, 'M', 'Ash ', 'VM', 'FC', 'FT', 'PS']]
    display(df_tetangga)
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer


# Daftarkan semua kolom numerik yang dibutuhkan oleh KNN untuk mengenali "sidik jari" biomassa
all_numerical_inputs = ['M', 'Ash ', 'VM', 'FC', 'C', 'H', 'O', 'N', 'HR', 'FR', 'PS']
op_features = ['PS']

print("==========================================================")
print("     PROSES IMPUTASI KNN IMPUTER BERBASIS SPESIES        ")
print("==========================================================")

# --- LANGKAH A: RINGKAS DATA MENJADI REKAP LEVEL SPESIES (GROUPBY) ---
# Trik utama: Kita satukan 13 baris Cherry seed menjadi 1 baris profil tunggal menggunakan nilai median
df_species_profile = df_final_PS[['Biomass species'] + all_numerical_inputs].groupby('Biomass species').median()

# --- LANGKAH B: JALANKAN KNN IMPUTER HANYA PADA PROFIL SPESIES TERSEBUT ---
# Dengan begini, KNN hanya menebak SATU angka keputusan untuk setiap nama spesies
knn_imputer_op = KNNImputer(n_neighbors=5)
df_species_imputed = pd.DataFrame(
    knn_imputer_op.fit_transform(df_species_profile),
    columns=all_numerical_inputs,
    index=df_species_profile.index
)
print("✔ KNN Imputer berhasil menentukan nilai logis per nama spesies.")

# --- LANGKAH C: MAP KEMBALI SEBAGAI GARANSI KESERAGAMAN ---
# Menembakkan angka tunggal hasil tebakan KNN tadi secara massal ke baris aslinya
for col in op_features:
    # Buat kamus pemetaan: {Nama Spesies: Nilai Tebakan KNN}
    mapping_dict = df_species_imputed[col].to_dict()

    # Isi hanya baris yang masih kosong (NaN) dengan kamus pemetaan di atas
    df_final_PS[col] = df_final_PS[col].fillna(df_final_PS['Biomass species'].map(mapping_dict))

print("✔ Pengisian selesai. Angka terkunci seragam untuk spesies yang sama!")
print("-" * 58)
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

# 1. Ambil data asli dari df_final_PS (Checkpoint aman)
df_HR_FR_2 = df_final_PS.copy()

# Tentukan sidik jari kimia utuh + kondisi termal yang sudah lengkap untuk pemandu KNN
all_numerical_inputs = ['M', 'Ash ', 'VM', 'FC', 'C', 'H', 'O', 'N', 'FT', 'PS', 'HR', 'FR']

print("====================================================================")
print("     EKSEKUSI FINAL HANDLING HR & FR (SINKRONISASI + KNN PINTAR)    ")
print("====================================================================")

# --- TAHAP 1: SINKRONISASI BARIS KEMBAR (FFILL & BFILL) ---
# Mengunci berdasarkan rumpun keluarga percobaan yang proksimat & ultimatnya kembar total
kolom_kunci_batch = ['Biomass species', 'M', 'Ash ', 'VM', 'FC', 'C', 'H', 'O', 'N']
df_HR_FR_2['HR'] = df_HR_FR_2.groupby(kolom_kunci_batch)['HR'].transform(lambda x: x.ffill().bfill())
df_HR_FR_2['FR'] = df_HR_FR_2.groupby(kolom_kunci_batch)['FR'].transform(lambda x: x.ffill().bfill())

# Simpan daftar nilai unik yang valid/nyata dari data asli untuk acuan pembulatan alat lab
nilai_valid_hr = df_final_PS['HR'].dropna().unique()
nilai_valid_fr = df_final_PS['FR'].dropna().unique()


# --- TAHAP 2: KNN IMPUTER UNTUK KELOMPOK YANG KOSONG SATU KAMPUNG ---
if df_HR_FR_2['HR'].isnull().any() or df_HR_FR_2['FR'].isnull().any():
    print("Mengisi sisa data kosong via KNN Imputer Terarah...")

    # Ambil matriks numerik input saja (SAMA SEKALI TIDAK ADA KOLOM YIELD)
    df_for_knn = df_HR_FR_2[all_numerical_inputs].copy()
    imputer = KNNImputer(n_neighbors=5)
    df_imputed = pd.DataFrame(imputer.fit_transform(df_for_knn), columns=all_numerical_inputs, index=df_HR_FR_2.index)

    # Fungsi pembulat mata uang/angka standar alat laboratorium terdekat
    def cari_angka_lab_terdekat(nilai_tebakan, daftar_nilai_valid):
        return min(daftar_nilai_valid, key=lambda x: abs(x - nilai_tebakan))

    # Masukkan hasil tebakan KNN yang sudah ditarik ke angka standar alat lab terdekat
    for idx in df_HR_FR_2[df_HR_FR_2['HR'].isnull()].index:
        tebakan_mentah = df_imputed.loc[idx, 'HR']
        df_HR_FR_2.loc[idx, 'HR'] = cari_angka_lab_terdekat(tebakan_mentah, nilai_valid_hr)

    for idx in df_HR_FR_2[df_HR_FR_2['FR'].isnull()].index:
        tebakan_mentah = df_imputed.loc[idx, 'FR']
        df_HR_FR_2.loc[idx, 'FR'] = cari_angka_lab_terdekat(tebakan_mentah, nilai_valid_fr)

print("✔ Seluruh data kosong pada input sukses ditangani secara logis!")
print("-" * 68)
```

```python
import pandas as pd
import numpy as np

# 1. Ambil kembali data asli dari checkpoint aman kamu (df_final_PS)
df_HR_FR_3 = df_final_PS.copy()

print("====================================================================")
print("    SOLUSI TOTAL: IMPUTASI BERDASARKAN TREN SEKUENSAL REAKTOR (LINEAR) ")
print("====================================================================")

# Urutkan data berdasarkan Spesies dan Tingkat Suhu FT agar polanya runtut secara fisis
df_HR_FR_3 = df_HR_FR_3.sort_values(by=['Biomass species', 'FT']).reset_index(drop=False)

# --- TAHAP 1: INTERPOLASI LINEAR BERDASARKAN TREN KELOMPOK SPESIES ---
# Metode ini akan membaca tren grafik naik/turunnya HR dan FR berdasarkan pergerakan suhu FT
df_HR_FR_3['HR'] = df_HR_FR_3.groupby('Biomass species')['HR'].transform(lambda x: x.interpolate(method='linear').ffill().bfill())
df_HR_FR_3['FR'] = df_HR_FR_3.groupby('Biomass species')['FR'].transform(lambda x: x.interpolate(method='linear').ffill().bfill())

# Kembalikan urutan indeks ke data aslinya agar posisi barismu tidak berantakan
df_HR_FR_3 = df_HR_FR_3.sort_values(by='Index').set_index('Index')

# --- TAHAP 2: PEMBULATAN STANDAR ALAT LAB (AGAR ANGKA TETAP BULAT RIIL) ---
nilai_valid_hr = df_final_PS['HR'].dropna().unique()
nilai_valid_fr = df_final_PS['FR'].dropna().unique()

def cari_angka_lab_terdekat(nilai_tebakan, daftar_nilai_valid):
    if pd.isna(nilai_tebakan): return nilai_tebakan
    return min(daftar_nilai_valid, key=lambda x: abs(x - nilai_tebakan))

df_HR_FR_3['HR'] = df_HR_FR_3['HR'].apply(lambda x: cari_angka_lab_terdekat(x, nilai_valid_hr))
df_HR_FR_3['FR'] = df_HR_FR_3['FR'].apply(lambda x: cari_angka_lab_terdekat(x, nilai_valid_fr))


# --- TAHAP 3: BACKUP AKHIR JIKA ADA SPESIES YANG KOSONG BLANK BLOK ---
# Jika ada satu jenis spesies yang dari awal TIDAK PUNYA data angka sama sekali untuk diinterpolasi,
# barulah diisi menggunakan modus global sebagai langkah penyelamatan terakhir.
if df_HR_FR_3['HR'].isnull().any():
    df_HR_FR_3['HR'] = df_HR_FR_3['HR'].fillna(df_final_PS['HR'].mode()[0])
if df_HR_FR_3['FR'].isnull().any():
    df_HR_FR_3['FR'] = df_HR_FR_3['FR'].fillna(df_final_PS['FR'].mode()[0])

print("✔ Proses rekonstruksi tren operasional laboratorium sukses dilakukan!")
print("-" * 68)

# --- VERIFIKASI AKHIR INPUT DATA ---
all_numerical_inputs = ['M', 'Ash ', 'VM', 'FC', 'C', 'H', 'O', 'N', 'FT', 'PS', 'HR', 'FR']
print("=== VERIFIKASI KEKOSONGAN INPUT (WAJIB ALL 0) ===")
print(df_HR_FR_3[all_numerical_inputs].isnull().sum())
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

# 1. Ambil data terakhir yang inputnya sudah bersih 100% via Cara 3
df_yield_final = df_HR_FR_3.copy()

# Daftarkan nama kolom yield target
kolom_yield = ['Solid phase', 'Liquid phase', 'Gas phase']
# Daftarkan seluruh fitur input numerik yang sudah lengkap sebagai jangkar pemandu fisis
fitur_input = ['M', 'Ash ', 'VM', 'FC', 'C', 'H', 'O', 'N', 'FT', 'PS', 'HR', 'FR']

print("====================================================================")
print("     PROSES PREPROCESSING YIELD: REKONSTRUKSI TEORI TEKNIK KIMIA     ")
print("====================================================================")

# --- TAHAP 1: URUTKAN DATA BERDASARKAN TREN TEMPERATUR FT ---
# Ini sengaja dilakukan untuk menyelamatkan 9 baris yang kosong total tiga-tiganya
# agar mereka terisi mengikuti kemiringan grafik suhu tetangga satu spesiesnya.
df_yield_final = df_yield_final.sort_values(by=['Biomass species', 'FT']).reset_index(drop=False)

for col in kolom_yield:
    df_yield_final[col] = df_yield_final.groupby('Biomass species')[col].transform(
        lambda x: x.interpolate(method='linear')
    )

# --- TAHAP 2: KNN IMPUTER TERARAH UNTUK 27 BARIS (SISA RASIO) ---
# Jika setelah interpolasi masih ada yang bolong (karena tidak ada jangkar atas-bawah),
# kita selesaikan dengan KNN yang membaca seluruh fitur input PLUS data Solid phase yang selamat.
semua_fitur_pemandu = fitur_input + kolom_yield

df_for_knn_yield = df_yield_final[semua_fitur_pemandu].copy()
imputer_yield = KNNImputer(n_neighbors=5)
df_imputed_yield = pd.DataFrame(
    imputer_yield.fit_transform(df_for_knn_yield),
    columns=semua_fitur_pemandu,
    index=df_yield_final.index
)

# Masukkan tebakan KNN murni hanya untuk mengisi yang masih kosong saja
for col in kolom_yield:
    df_yield_final[col] = df_yield_final[col].fillna(df_imputed_yield[col])


# --- TAHAP 3: KEMBALIKAN INDEKS & NORMALISASI TOTAL MASSA MAKSIMAL 100% ---
df_yield_final = df_yield_final.sort_values(by='Index').set_index('Index')

# Paksa hitungan matematika agar total Solid + Liquid + Gas pas 100% (Hukum Kekekalan Massa)
total_massa_per_baris = df_yield_final[kolom_yield].sum(axis=1)
for col in kolom_yield:
    df_yield_final[col] = (df_yield_final[col] / total_massa_per_baris) * 100

print("✔ Tahap 1 & 2: Interpolasi Tren Suhu + KNN Sisa Rasio sukses.")
print("✔ Tahap 3: Normalisasi Hukum Kekekalan Massa (Total = 100%) sukses terkunci.")
print("-" * 72)

# --- VERIFIKASI TOTAL AKHIR ---
print("=== AUDIT FINAL DATASET (WAJIB ALL 0 KOSONG) ===")
print(df_yield_final[fitur_input + kolom_yield].isnull().sum())
print(f"\n🎉 DATA SELESAI: {len(df_yield_final)} baris data kini padat, logis fisis, dan siap modeling!")
```

```python
for col in ['Solid phase', 'Liquid phase', 'Gas phase']:
    df_clean[col] = df_clean[col].astype(str).str.extract(r'(\d+\.?\d*)')[0]
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

def clean_ps_to_float(val):
    if pd.isnull(val) or val == 'nan':
        return np.nan

    val_str = str(val).strip()

    # Jika mendeteksi pola rentang dengan tanda strip (misal '0.5-0.6')
    if '-' in val_str:
        try:
            parts = val_str.split('-')
            # Hitung nilai tengahnya: (0.5 + 0.6) / 2 = 0.55
            return (float(parts[0]) + float(parts[1])) / 2.0
        except:
            return np.nan

    # Jika angka biasa, langsung ubah ke float
    try:
        return float(val_str)
    except:
        return np.nan

# Terapkan fungsi pembersihan pada kolom PS
df_clean['PS_numeric'] = df_clean['PS'].apply(clean_ps_to_float)

print("Tipe data PS sekarang:", df_clean['PS_numeric'].dtype)
print("Jumlah missing value di PS sekarang:", df_clean['PS_numeric'].isnull().sum())
```

# PIPELINE CLEANING

```python
import numpy as np
import pandas as pd
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.impute import KNNImputer

class BiomassThermochemicalImputer(BaseEstimator, TransformerMixin):
    def __init__(self, n_neighbors=5):
        self.n_neighbors = n_neighbors
        self.prox_cols = ['M', 'Ash ', 'VM', 'FC']
        self.ult_cols = ['C', 'H', 'O', 'N']
        self.all_features = ['M', 'Ash ', 'VM', 'FC', 'C', 'H', 'O', 'N', 'HR', 'FR']

        # State Data Train (Proksimat & Ultimat)
        self.median_prox_ = {}
        self.median_ult_ = {}
        self.knn_imputer_prox_ = None
        self.knn_imputer_ult_ = None

        # State Data Train (PS)
        self.knn_imputer_ps_ = None
        self.species_ps_mapping_dict_ = {}

        # State Data Train (HR & FR - Tren Laboratorium)
        self.valid_hr_values_ = None
        self.valid_fr_values_ = None
        self.global_mode_hr_ = None
        self.global_mode_fr_ = None

    def fit(self, X, y=None):
        X_df = pd.DataFrame(X).copy()
        species_col = 'Biomass species' if 'Biomass species' in X_df.columns else X_df.columns[1]

        # 1. FIT PROKSIMAT
        empat_kosong_mask = X_df[self.prox_cols].isnull().all(axis=1)
        df_normal_prox = X_df[~empat_kosong_mask]
        for col in self.prox_cols:
            self.median_prox_[col] = df_normal_prox[col].median()

        self.knn_imputer_prox_ = KNNImputer(n_neighbors=self.n_neighbors)
        avail_numeric = [c for c in self.all_features if c in X_df.columns and X_df[c].dtype != 'object']
        self.knn_imputer_prox_.fit(X_df[avail_numeric])

        # 2. FIT ULTIMAT
        total_ult_sementara = X_df[self.ult_cols].sum(axis=1)
        anomali_ult_mask = total_ult_sementara > 100.0
        df_normal_ult = X_df.copy()
        df_normal_ult.loc[anomali_ult_mask, self.ult_cols] = np.nan
        ult_kosong_semua = df_normal_ult[self.ult_cols].isnull().all(axis=1)

        df_valid_ult = df_normal_ult[~ult_kosong_semua]
        for col in self.ult_cols:
            self.median_ult_[col] = df_valid_ult[col].median()

        self.knn_imputer_ult_ = KNNImputer(n_neighbors=self.n_neighbors)
        self.knn_imputer_ult_.fit(X_df[avail_numeric])

        # 3. FIT PARTICLE SIZE (PS) - Mengambil sisa profil jika masih ada yang kosong murni
        if 'PS' in X_df.columns:
            all_knn_ps_features = self.all_features + ['PS']
            df_species_profile = X_df[[species_col] + all_knn_ps_features].groupby(species_col).median()

            self.knn_imputer_ps_ = KNNImputer(n_neighbors=self.n_neighbors)
            df_species_imputed = pd.DataFrame(
                self.knn_imputer_ps_.fit_transform(df_species_profile),
                columns=all_knn_ps_features,
                index=df_species_profile.index
            )
            self.species_ps_mapping_dict_ = df_species_imputed['PS'].to_dict()

        # 4. FIT OPERATIONAL TREND (HR & FR)
        if 'HR' in X_df.columns:
            self.valid_hr_values_ = X_df['HR'].dropna().unique()
            self.global_mode_hr_ = X_df['HR'].mode()[0] if not X_df['HR'].mode().empty else 10.0
        if 'FR' in X_df.columns:
            self.valid_fr_values_ = X_df['FR'].dropna().unique()
            self.global_mode_fr_ = X_df['FR'].mode()[0] if not X_df['FR'].mode().empty else 1.0

        return self

    def _cari_angka_lab_terdekat(self, nilai_tebakan, daftar_nilai_valid):
        if pd.isna(nilai_tebakan) or daftar_nilai_valid is None or len(daftar_nilai_valid) == 0:
            return nilai_tebakan
        return min(daftar_nilai_valid, key=lambda x: abs(x - nilai_tebakan))

    def transform(self, X):
        X_df = pd.DataFrame(X).copy()
        species_col = 'Biomass species' if 'Biomass species' in X_df.columns else X_df.columns[1]

        if 'Index' not in X_df.columns:
            X_df = X_df.reset_index().rename(columns={'index': 'Index'})

        # ==========================================
        # TAHAP 1: CLEANING PROKSIMAT
        # ==========================================
        empat_kosong_mask = X_df[self.prox_cols].isnull().all(axis=1)
        df_normal = X_df[~empat_kosong_mask].copy()
        df_blanks = X_df[empat_kosong_mask].copy()

        for idx, row in df_normal.iterrows():
            null_cols = [col for col in self.prox_cols if pd.isnull(row[col])]
            if len(null_cols) == 0: continue
            dependent_col = null_cols[-1]
            cols_to_fill_with_median = null_cols[:-1]
            for col in cols_to_fill_with_median:
                df_normal.at[idx, col] = self.median_prox_[col]
            current_sum = sum(df_normal.at[idx, c] for c in self.prox_cols if c != dependent_col)
            df_normal.at[idx, dependent_col] = max(0.0, 100.0 - current_sum)

        if len(df_blanks) > 0:
            avail_numeric = [c for c in self.all_features if c in X_df.columns and X_df[c].dtype != 'object']
            knn_imputed = pd.DataFrame(self.knn_imputer_prox_.transform(X_df[avail_numeric]),
                                       columns=avail_numeric, index=X_df.index)
            for col in self.prox_cols:
                df_blanks[col] = knn_imputed.loc[empat_kosong_mask, col]
            for idx, row in df_blanks.iterrows():
                total_knn = sum(df_blanks.at[idx, col] for col in self.prox_cols)
                if total_knn > 0:
                    for col in self.prox_cols:
                        df_blanks.at[idx, col] = (df_blanks.at[idx, col] / total_knn) * 100.0

        df_prox_clean = pd.concat([df_normal, df_blanks]).sort_index()
        total_asli_prox = df_prox_clean[self.prox_cols].sum(axis=1)
        for idx, row in df_prox_clean.iterrows():
            t_baris = total_asli_prox.loc[idx]
            if t_baris <= 0:
                for col in self.prox_cols: df_prox_clean.at[idx, col] = self.median_prox_[col]
            else:
                for col in self.prox_cols: df_prox_clean.at[idx, col] = (df_prox_clean.at[idx, col] / t_baris) * 100.0

        # ==========================================
        # TAHAP 2: CLEANING ULTIMAT
        # ==========================================
        df_ult = df_prox_clean.copy()
        total_ult_sementara = df_ult[self.ult_cols].sum(axis=1)
        anomali_mask = total_ult_sementara > 100.0
        df_ult.loc[anomali_mask, self.ult_cols] = np.nan

        ult_kosong_semua = df_ult[self.ult_cols].isnull().all(axis=1)
        df_ult_normal = df_ult[~ult_kosong_semua].copy()
        df_ult_blanks = df_ult[ult_kosong_semua].copy()

        for idx, row in df_ult_normal.iterrows():
            null_cols_ult = [col for col in self.ult_cols if pd.isnull(row[col])]
            if len(null_cols_ult) == 0: continue
            dependent_col_ult = null_cols_ult[-1]
            cols_to_fill_with_median_ult = null_cols_ult[:-1]
            for col in cols_to_fill_with_median_ult: df_ult_normal.at[idx, col] = self.median_ult_[col]
            current_sum_ult = sum(df_ult_normal.at[idx, c] for c in self.ult_cols if c != dependent_col_ult)
            df_ult_normal.at[idx, dependent_col_ult] = max(0.0, 100.0 - current_sum_ult)

        if len(df_ult_blanks) > 0:
            avail_numeric = [c for c in self.all_features if c in df_ult.columns and df_ult[c].dtype != 'object']
            knn_imputed_ult = pd.DataFrame(self.knn_imputer_ult_.transform(df_ult[avail_numeric]),
                                           columns=avail_numeric, index=df_ult.index)
            for col in self.ult_cols: df_ult_blanks[col] = knn_imputed_ult.loc[ult_kosong_semua, col]
            for idx, row in df_ult_blanks.iterrows():
                t_baris_blank = sum(df_ult_blanks.at[idx, col] for col in self.ult_cols)
                if t_baris_blank > 0:
                    for col in self.ult_cols: df_ult_blanks.at[idx, col] = (df_ult_blanks.at[idx, col] / t_baris_blank) * 100.0

        df_final_clean = pd.concat([df_ult_normal, df_ult_blanks]).sort_index()
        for col in self.ult_cols: df_final_clean[col] = df_final_clean[col].clip(lower=0.0)

        # ==========================================
        # TAHAP 3: HANDLING FITUR PS (SISA KOSONG TOTAL)
        # ==========================================
        if 'PS' in df_final_clean.columns:
            df_final_clean['PS'] = df_final_clean['PS'].fillna(df_final_clean[species_col].map(self.species_ps_mapping_dict_))
            if df_final_clean['PS'].isnull().any():
                global_median_ps = np.nanmedian(list(self.species_ps_mapping_dict_.values()))
                df_final_clean['PS'] = df_final_clean['PS'].fillna(global_median_ps)

        # ==========================================
        # TAHAP 4: HANDLING HR & FR (LOGIKA TREN LINEAR)
        # ==========================================
        df_trend = df_final_clean.copy()
        df_trend = df_trend.sort_values(by=[species_col, 'FT']).reset_index(drop=True)

        df_trend['HR'] = df_trend.groupby(species_col)['HR'].transform(lambda x: x.interpolate(method='linear').ffill().bfill())
        df_trend['FR'] = df_trend.groupby(species_col)['FR'].transform(lambda x: x.interpolate(method='linear').ffill().bfill())

        if df_trend['HR'].isnull().any():
            df_trend['HR'] = df_trend['HR'].fillna(self.global_mode_hr_)
        if df_trend['FR'].isnull().any():
            df_trend['FR'] = df_trend['FR'].fillna(self.global_mode_fr_)

        df_trend['HR'] = df_trend['HR'].apply(lambda x: self._cari_angka_lab_terdekat(x, self.valid_hr_values_))
        df_trend['FR'] = df_trend['FR'].apply(lambda x: self._cari_angka_lab_terdekat(x, self.valid_fr_values_))

        df_trend = df_trend.sort_values(by='Index').set_index('Index')
        return df_trend
```

```python
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import cross_validate, KFold
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor

# ====================================================================
# 1. LOAD DATASET DF_TRAIN YANG SUDAH SINKRON UTUH
# ====================================================================
path_df_train = '/content/drive/MyDrive/SEM 6/02. AI TERAPAN/dataset_pyrolysis/cleaning/cleaning_revisi/df_train.csv'

# Baca dengan mengunci kolom 'Index' agar selaras secara internal
df_train = pd.read_csv(path_df_train, index_col='Index')

# ====================================================================
# 2. SEPARASI SUMBU X (FITUR) DAN SUMBU Y (TARGET)
# ====================================================================
# Tentukan nama-nama kolom target yield
kolom_yield = ['Solid phase', 'Liquid phase', 'Gas phase']

# Buat matriks input X dengan membuang kolom target dan kolom non-numerik pembantu
kolom_dibuang = ['Biomass species', 'index', 'Unnamed: 0'] + kolom_yield
X = df_train.drop(columns=kolom_dibuang, errors='ignore')

# Ambil target yield langsung dari baris yang sama di df_train
target_yields = {
    'SOLID PHASE (Char)': df_train['Solid phase'],
    'LIQUID PHASE (Bio-oil)': df_train['Liquid phase'],
    'GAS PHASE': df_train['Gas phase']
}

# ====================================================================
# 3. KONFIGURASI ALGORITMA DAN STRATEGI VALIDASI (5-FOLD CV)
# ====================================================================
models = {
    'Random Forest': RandomForestRegressor(random_state=42),
    'XGBoost': XGBRegressor(random_state=42),
    'LightGBM': LGBMRegressor(random_state=42, verbose=-1)
}

# KFold dengan shuffle=True menjamin pengacakan baris selama CV berjalan aman
# karena X dan y Anda saat ini sudah berbagi indeks baris yang sama secara absolut.
kfold = KFold(n_splits=5, shuffle=True, random_state=42)

# ====================================================================
# 4. EKSEKUSI EVALUASI PER KELOMPOK TARGET YIELD
# ====================================================================
print("====================================================================")
print("     FASE EVALUASI: SPOT-CHECKING MULTI-ALGORITMA PER TARGET YIELD  ")
print("          (MENGGUNAKAN DATAFRAME DF_TRAIN YANG SUDAH SINKRON)       ")
print("====================================================================")
print(f"Dimensi Matriks Fitur Input X: {X.shape}")

for nama_target, y in target_yields.items():
    print(f"\n🎯 TARGET EVALUASI: {nama_target}")
    print("=" * 60)

    for nama_model, model in models.items():
        # Hitung validasi silang murni untuk melihat performa asli algoritma
        skor = cross_validate(model, X, y, cv=kfold, scoring=['r2', 'neg_mean_absolute_error'])

        r2_rata2 = np.mean(skor['test_r2'])
        mae_rata2 = np.mean(-skor['test_neg_mean_absolute_error'])

        print(f"🤖 Model: {nama_model:<15} | R² Score: {r2_rata2:.4f} | MAE Error: {mae_rata2:.4f} %")

    print("-" * 60)
```
